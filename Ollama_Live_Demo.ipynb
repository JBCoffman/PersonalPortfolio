# Local LLM Setup via Ollama Demo

## Begin with environmental dependencies.

###**Step 1:** Select `Runtime` in the menu above, choose `Change runtime type` and set `Hardware accellerator` to `T4 GPU`

### **Step 2:** Run the code block below to install Ollama's system package dependency to demo environment.

```python
!apt-get install -y zstd
```

### Now to install Ollama itself, click the code block below.

```python
!curl -fsSL https://ollama.com/install.sh | sh
!nohup ollama serve > ollama.log 2>&1 &
```

### Now to verify proper install and Ollama is ready to receive requests run code below. Look for "Listening on..."

```python
!sleep 2 && cat ollama.log
```

### Ollama is now ready, so let's pull down a basic model to test. Full list of models available at https://ollama.com/library

```python
!sleep 3 && ollama pull llama3.2:3b
```

### Once model is pulled down, lets run a test query below. Feel free to modify request input.

*With lower end hardware, this can take a while to return a response.*

```python
!ollama run llama3.2:3b "Explain what makes local LLMs different from cloud LLMs, in two sentences"
```
